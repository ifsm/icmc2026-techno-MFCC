```python
import glob
import os
import numpy as np
import pandas as pd
from pathlib import Path

# Verzeichnisse
measCsvMFCC = r'C:\Users\t_ziemer\Desktop\techno-mfcc'
foldsMFCC = glob.glob(str(measCsvMFCC) + '\\*')

measCsvBPM = r'C:\Users\t_ziemer\Desktop\techno-csv'
foldsBPM = glob.glob(str(measCsvBPM) + '\\*')

measCsvMFCCBPM = r'C:\Users\t_ziemer\Desktop\techno-mfccbpm'
Path(measCsvMFCCBPM).mkdir(parents=True, exist_ok=True)

for folder in foldsMFCC:
    folder_name = os.path.basename(folder)
    bpm_folder = os.path.join(measCsvBPM, folder_name)
    output_folder = os.path.join(measCsvMFCCBPM, folder_name)
    Path(output_folder).mkdir(parents=True, exist_ok=True)

    mfcc_files = glob.glob(os.path.join(folder, '*.csv'))

    for mfcc_file in mfcc_files:
        filename = os.path.basename(mfcc_file)
        bpm_file = os.path.join(bpm_folder, filename)

        if os.path.exists(bpm_file):
            # CSV laden, index_col=0 um die Index-Spalte zu ignorieren (leeres Feld am Anfang)
            mfcc_df = pd.read_csv(mfcc_file, header=0, index_col=0)
            line_numbers = mfcc_df.index
            mfcc_array = mfcc_df.values

            bpm_df = pd.read_csv(bpm_file, header=0)
            bpm_column = bpm_df.iloc[:, -1].values

            # Zeilen kürzen, um gleiche Länge zu haben
            min_length = min(mfcc_array.shape[0], bpm_column.shape[0])
            mfcc_array = mfcc_array[:min_length, :]
            bpm_cropped = bpm_column[:min_length]
            line_numbers = line_numbers[:min_length]

            # Neue Spalte anhängen
            combined_array = np.column_stack((mfcc_array, bpm_cropped))

            # Kopfzeilen extrahieren
            original_headers = list(mfcc_df.columns)
            # Die erste Spalte ist bereits die Index-Spalte (leeres Header-Feld)
            headers = [''] + original_headers + ['bpm']

            # DataFrame erstellen
            combined_df = pd.DataFrame(combined_array, columns=headers[1:])
            # Zeilennummern wieder hinzufügen (Index)
            combined_df.insert(0, '', line_numbers)

            # In CSV speichern, Index wird automatisch mit geschrieben, weil wir ihn reingefügt haben
            output_path = os.path.join(output_folder, filename)
            combined_df.to_csv(output_path, index=False)
        else:
            print(f"Keine passende BPM-Datei für {filename} gefunden.")
```
